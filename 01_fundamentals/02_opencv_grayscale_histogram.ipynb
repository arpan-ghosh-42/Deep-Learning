```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Read the color image
image = cv2.imread("sample.jpg")

# Check whether image is loaded
if image is None:
    print("Image not found!")
else:
    # Step 2: Display original color image
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(5, 5))
    plt.imshow(image_rgb)
    plt.title("Original Color Image")
    plt.axis("off")
    plt.show()

    # Step 3: Convert color image to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Step 4: Display grayscale image
    plt.figure(figsize=(5, 5))
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.show()

    # ------------------------------------------------
    # HISTOGRAM USING BUILT-IN FUNCTION
    # ------------------------------------------------

    # Step 5: Calculate histogram using function
    histogram = cv2.calcHist([gray], [0], None, [256], [0, 256])

    # Step 6: Plot histogram
    plt.figure(figsize=(8, 4))
    plt.plot(histogram)
    plt.title("Histogram Using Function")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Frequency")
    plt.xlim([0, 256])
    plt.show()

    # ------------------------------------------------
    # HISTOGRAM WITHOUT USING HISTOGRAM FUNCTION
    # ------------------------------------------------

    # Step 7: Create 256 bins and initialize with 0
    manual_histogram = [0] * 256

    # Step 8: Traverse every pixel
    rows, columns = gray.shape

    for i in range(rows):
        for j in range(columns):

            # Get pixel intensity
            intensity = gray[i, j]

            # Increase frequency
            manual_histogram[intensity] += 1

    # Step 9: Plot manually calculated histogram
    plt.figure(figsize=(8, 4))
    plt.plot(range(256), manual_histogram)
    plt.title("Histogram Without Using Function")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Frequency")
    plt.xlim([0, 256])
    plt.show()
```
